```python
import glob
import time

print("Initalizing classification script...")

file_exists = os.path.exists("outputs/final/github_repos_classified.csv")
print("File exists:", file_exists)

if not file_exists:
    print("Creating new classification file...")
    repos_df = pd.read_csv("outputs/final/github_repos_all_parts.csv")
    repos_df['classified'] = pd.Series(dtype='string')
    repos_df.to_csv("outputs/final/github_repos_classified.csv", index=False)

repos_df = pd.read_csv("outputs/final/github_repos_classified.csv")

repos_df['classified'] = repos_df['classified'].astype("string")

# # Shuffle the dataframe
# repos_df = repos_df.sample(frac=1).reset_index(drop=True)

for idx, row in repos_df.iterrows():
    # Some of the test sets I've tried using
    # keywords = ["biology", "physics", "computer science", "research", "publish", "University", "scientific"]
    # keywords = ["academic", "academia", "professor", "phd", "postdoc"]
    # keywords = ["@article", "@inproceedings", "@misc", "doi=", "Associated Publication", "Published in", "Our paper", "preprint"]
    science_terms = [
        "science", "research", "data", "dataset", "publication", "doi", "@article{", "pmc",
        "genome", "genomics", "phage", "tcr", "aging", "muscle", "bioinformatics",
        "medicine", "health", "clinical", "open science"
    ]
    science_terms = ["doi.org", "doi:", "doi =", "doi=", "@article{", "@inproceedings{", "@misc{", "pmc", "pmid", "cite", "our findings", "our paper", "published in", "associated publication", "science", "research"]
    key = False
    if pd.notna(row["readme"]):
        for kw in science_terms:
            if kw in row["readme"]:
                key = True
                break
    else:
        key = False
    if pd.isna(row['classified']) and (key and row['language'] not in ['HTML', 'CSS']) and "ml-latest-small" not in row["readme"] and "methodology" in row["readme"]:
        print(f"Repo: {row['html_url']}")
        time.sleep(0.5)
        answer = input("Classify as 'yes', 'maybe' or 'no': ").strip().lower()
        if answer in ['yes', 'maybe', 'no']:
            repos_df.at[idx, 'classified'] = answer
        else:
            print("Ending classification.")
            break
    else:
        continue

# Save the modified dataframe
print("Final classification counts:")
print(repos_df['classified'].value_counts(dropna=False))
repos_df.to_csv("outputs/final/github_repos_classified.csv", index=False)
```
